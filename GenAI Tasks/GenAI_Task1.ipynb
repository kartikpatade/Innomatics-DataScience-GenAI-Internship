# Mini Prompt Engine with LangChain

Converting user input into structured prompts using LangChain templates. All tasks implemented with `PromptTemplate` and `ChatPromptTemplate`.

```python
import langchain
from langchain.prompts import PromptTemplate, ChatPromptTemplate

ALLOWED_AUDIENCES = {"beginner", "intermediate", "expert"}
ALLOWED_TONES = {"formal", "casual", "fun"}
ALLOWED_STYLES = {"teaching", "interview", "storytelling"}
print("LangChain version:", langchain.__version__)
```

## Task 1: Replace Hardcoded Prompts with PromptTemplate

Convert `f"Explain {topic} in simple terms for beginners"` into a reusable `PromptTemplate`.

```python
explain_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

def explain_topic(topic):
    """Return a beginner-friendly explanation prompt."""
    return explain_template.format(topic=topic)

print("Task 1 Output:")
print(explain_topic("AI"))
```

## Task 2: Multi-Input Prompt System

Build a template with inputs: `topic`, `audience`, `tone`

```python
multi_input_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

test_cases = [
    {"topic": "AI", "audience": "beginners", "tone": "friendly"},
    {"topic": "Python", "audience": "kids", "tone": "fun"},
    {"topic": "Deep Learning", "audience": "engineers", "tone": "technical"}
]

print("Task 2 Outputs:")
for i, case in enumerate(test_cases, 1):
    print(f"Test case {i}: {multi_input_template.format(**case)}")
```

## Task 3: Prompt Variations Engine

Create 3 different prompt templates: Teaching, Interview, and Storytelling.

```python
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)

storytelling_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

variation_templates = {
    "Teaching": teaching_template,
    "Interview": interview_template,
    "Storytelling": storytelling_template
}

topic_variation = "Machine Learning"
print("Task 3 Outputs:")
for name, template in variation_templates.items():
    print(f"{name}: {template.format(topic=topic_variation)}")
```

## Task 4: ChatPromptTemplate System

Build chat-based prompts with system and user roles.

```python
chat_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a {role}. Respond in that style."),
        ("human", "Topic: {topic}")
    ]
)

role_value = "teacher"
topic_value = "Neural Networks"
messages = chat_prompt.format_messages(role=role_value, topic=topic_value)

print("Task 4 Output:")
for message in messages:
    print(f"{message.type.upper()}: {message.content}")
```

## Task 5: Input Validation Layer

Validate `audience` and `tone` against predefined allowed values.

```python
def validate_inputs(audience, tone):
    """Validate audience and tone against allowed values."""
    if audience not in ALLOWED_AUDIENCES:
        raise ValueError(f"Invalid audience: {audience}. Allowed: {ALLOWED_AUDIENCES}")
    if tone not in ALLOWED_TONES:
        raise ValueError(f"Invalid tone: {tone}. Allowed: {ALLOWED_TONES}")
    return audience, tone

valid_audience, valid_tone = validate_inputs("beginner", "formal")
print("Task 5 Output:")
print(f"Validated: audience='{valid_audience}', tone='{valid_tone}'")

try:
    validate_inputs("invalid", "formal")
except ValueError as e:
    print(f"Error caught: {e}")
```

## Task 6: Prompt Generator App

Unified function combining validation and prompt generation.

```python
generator_template = PromptTemplate(
    input_variables=["topic", "audience", "tone", "style"],
    template="Explain {topic} for {audience} in a {tone} {style} style"
)

def generate_prompt(topic, audience, tone, style):
    """Generate a prompt using validation and a reusable template."""
    audience, tone = validate_inputs(audience, tone)
    if style not in ALLOWED_STYLES:
        raise ValueError(f"Invalid style: {style}. Allowed: {ALLOWED_STYLES}")
    return generator_template.format(
        topic=topic,
        audience=audience,
        tone=tone,
        style=style
    )

print("Task 6 Output:")
result = generate_prompt("Neural Networks", "beginner", "fun", "storytelling")
print(result)
```

Demonstrate reusability by using one template with 5 different inputs.

```python
reusability_cases = [
    {"topic": "Data Science", "audience": "beginner", "tone": "casual"},
    {"topic": "APIs", "audience": "intermediate", "tone": "formal"},
    {"topic": "Cloud Computing", "audience": "expert", "tone": "formal"},
    {"topic": "Cybersecurity", "audience": "beginner", "tone": "fun"},
    {"topic": "Databases", "audience": "intermediate", "tone": "casual"}
]

print("Task 7 Outputs (Reusability Test - Same Template, Different Inputs):")
print(f"\nUsing template: {multi_input_template.template}\n")
for i, case in enumerate(reusability_cases, 1):
    output = multi_input_template.format(**case)
    print(f"{i}. {output}")
```
