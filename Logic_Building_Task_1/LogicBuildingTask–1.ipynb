Task 1: User Login Check

```python
correct_user = "admin"  # Stores the correct username
correct_pass = "1234"   # Stores the correct password

enteredUser = "admin"   # Username entered by the user
enteredPass = "1234"    # Password entered by the user

if enteredUser == correct_user and enteredPass == correct_pass:  # Checks if both username and password match
    print("Login Successful")
else:  # Runs if username or password is wrong
    print("Invalid Credentials")
```

Task 2: Pass / Fail Analyzer

```python
student_marks = [45, 78, 90, 33, 60]  # List of student marks

passCount = 0  # Counter for passed
failCount = 0  # Counter for failed

for mark in student_marks:  # Loop through each student's mark
    if mark >= 50:  # Passing condition: marks must be 50 or more
        passCount = passCount + 1
    else:  # If marks are less than 50, student fails
        failCount += 1

print("Total Pass Students:", passCount)
print("Total Fail Students:", failCount)
```

Task 3: Simple Data Cleaner

```python
raw_names = [" Alice ", "bob", " CHARLIE "]

cleanNames = []  # List to store cleaned names

for nm in raw_names:  # Loop through each name in the list
    trimmed = nm.strip()  # Remove extra spaces from start and end
    lowered = trimmed.lower()  # Convert name to lowercase
    cleanNames.append(lowered)  # Add cleaned name to the list

print("Cleaned List:", cleanNames)
```

Task 4: Message Length Analyzer

```python
msgs = ["Hi", "Welcome to the platform", "OK"]

for message in msgs:  # Loop through each message
    msg_length = len(message)  # Find the length of the message

    print("Message:", "'" + message + "'", "| Length:", msg_length)  # Print message with its length

    if msg_length > 10:  # Flag messages longer than 10 characters
        print("Flag: Message longer than 10 characters")
    else:
        pass  # No action needed for short messages
```

Task 5: Error Message Detector

```python
system_logs = ["INFO", "ERROR", "WARNING", "ERROR"]

errorCounter = 0  # Counter to store total number of ERROR logs

for entry in system_logs:  # Loop through each log entry
    if entry == "ERROR":  # Check if the log type is ERROR (case-sensitive)
        errorCounter = errorCounter + 1
    else:
        continue  # Skip other log types

print("Total ERROR entries:", errorCounter)  # Print final error count
```
