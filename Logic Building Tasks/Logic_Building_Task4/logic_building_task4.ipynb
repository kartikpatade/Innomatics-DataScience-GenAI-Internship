## Problem 1: Mobile Recharge Validation System

```python
def validateRecharge():
    plans=[199,299,399,599]  # Valid recharge plans

    while True:
        try:
            amt=int(input("Enter recharge amount: "))
        except:
            print("Please enter numbers only.")
            continue

        if amt<50:
            print("Invalid: Recharge must be at least ₹50.")
        elif amt not in plans:
            print("Invalid plan. Available:",plans)
        else:
            print("Recharge Successful for ₹",amt)
            break  # Stop retry loop after successful recharge


validateRecharge()
```

## Problem 2: Inventory Reorder Alert System

```python
def inventoryCheck(prod_dict):
    for name in prod_dict:  # Loop through each product
        stock=prod_dict[name]

        if stock<15:
            print(name,"-> Reorder Alert (Stock:",stock,")")
        else:
            print(name,"-> Stock OK (Stock:",stock,")")


products_data={
    "Rice":10,
    "Sugar":25,
    "Oil":12,
    "Wheat":30
}

inventoryCheck(products_data)
```

## Problem 3: Student Result Processing System

```python
def studentResult(marks_list):
    totalMarks=0  # Stores total marks

    for m in marks_list:
        totalMarks=totalMarks+m  # Add each mark to total

    avg=totalMarks/len(marks_list)  # Calculate average

    if avg>=50:
        return "Pass"
    else:
        return "Fail"


marks=[45,60,70,50]
resultStatus=studentResult(marks)

print("Marks:",marks)
print("Result:",resultStatus)
```

## Problem 4: Cab Fare Estimator with Retry

```python
def calculateFare(distance,peak):
    baseFare=50
    perKmRate=12

    total=baseFare+(distance*perKmRate)

    if peak=="yes":
        total=total+(total*0.25)  # Add 25% extra during peak hour

    return total


while True:
    km=float(input("Enter distance in km: "))
    peakHour=input("Is it peak hour? (yes/no): ").lower()

    fareAmount=calculateFare(km,peakHour)
    print("Total Fare: ₹",fareAmount)

    again=input("Do you want to calculate again? (yes/no): ").lower()
    if again!="yes":
        break
```

## Problem 5: Employee Attendance Eligibility Checker

```python
def checkAttendance(att_list):
    presentDays=0  # Count of present days

    for d in att_list:
        if d=="P":
            presentDays+=1

    percent=(presentDays/len(att_list))*100  # Attendance percentage

    if percent>=75:
        return "Eligible"
    else:
        return "Not Eligible"


attendance_record=["P","P","A","P","P","A","P","P"]
eligibility=checkAttendance(attendance_record)

print("Attendance:",attendance_record)
print("Status:",eligibility)
```

## Problem 6: Password Strength Checker

```python
def checkPasswordStrength(pwd):
    hasDigit=False
    hasSpecial=False
    specials="@#$"

    if len(pwd)<8:
        return "Weak Password (minimum length is 8)"

    for ch in pwd:
        if ch.isdigit():
            hasDigit=True
        if ch in specials:
            hasSpecial=True

    if hasDigit and hasSpecial:
        return "Strong Password"
    else:
        return "Weak Password (must contain digit and special char @#$)"


userPwd=input("Enter password: ")
strength=checkPasswordStrength(userPwd)
print(strength)
```
