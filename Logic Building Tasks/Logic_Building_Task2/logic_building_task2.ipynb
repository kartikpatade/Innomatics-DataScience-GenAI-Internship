## Problem 1: Unique Words in a Sentence

```python
print("Problem 1: Unique Words in a Sentence\n")

sentence="python is easy and python is powerful"
words=sentence.split()
unique_words=set(words)

print("Unique words count:",len(unique_words))
print("Unique words:",unique_words)
```

## Problem 2: Highest Salary from Employee Data

```python
print("\nProblem 2: Highest Salary from Employee Data\n")

employees={
    "Ravi":75000,
    "Anita":68000,
    "Kiran":72000
}

highest_employee=max(employees,key=employees.get)
highest_salary=employees[highest_employee]

print("Highest Salary:",highest_employee,"-",highest_salary)
```

## Problem 3: Find Maximum and Minimum Values

```python
print("\nProblem 3: Find Maximum and Minimum Values\n")

numbers=[45,22,89,10,66]

max_value=numbers[0]
min_value=numbers[0]

for num in numbers:
    if num>max_value:
        max_value=num
    if num<min_value:
        min_value=num

print("List:",numbers)
print("Max:",max_value)
print("Min:",min_value)
```

## Problem 4: Count Products Above a Price Threshold

```python
print("\nProblem 4: Count Products Above a Price Threshold\n")

prices=[450,1200,899,1500,300]
count=0

for price in prices:
    if price>1000:
        count+=1

print("Products above 1000:",count)
```

## Problem 5: Calculate Attendance Percentage

```python
print("\nProblem 5: Calculate Attendance Percentage\n")

attendance=["P","P","A","P","P"]
present_count=0

for status in attendance:
    if status=="P":
        present_count+=1

percentage=(present_count/len(attendance))*100
print("Attendance Percentage:",percentage)
```

## Problem 6: Remove Duplicate Phone Numbers

```python
print("\nProblem 6: Remove Duplicate Phone Numbers\n")

phone_numbers=[9876543210,9123456789,9876543210]
unique_numbers=set(phone_numbers)

print("Unique phone numbers:",unique_numbers)
```

## Problem 7: Count Character Frequency

```python
print("\nProblem 7: Count Character Frequency\n")

text="python"
freq={}

for ch in text:
    freq[ch]=freq.get(ch,0)+1

print(freq)
```

## Problem 9: Check if a Key Exists in a Dictionary

```python
print("\nProblem 9: Check if a Key Exists in a Dictionary\n")

employees={
    "Ravi":75000,
    "Anita":68000,
    "Kiran":72000
}

key_to_check="Ravi"

if key_to_check in employees:
    print("Employee exists")
else:
    print("Employee does not exist")
```
